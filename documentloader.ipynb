```python
## Data Ingestion

# Document data structure

from langchain_core.documents import Document
import os
```

```python
doc1 = Document(
    page_content= "This is main RAG file document",
    metadata = {
        "source" : "example.txt",
        "pages" : 1,
        "author" : "Neelam"
    }
)
doc1

# Get the directory where this Python script is located
project_path = os.path.dirname(os.path.abspath(os.getcwd()))

# Build the path to the PDF folder
data_folder_path = os.path.join(
    project_path,
    "RAG",
    "data",
   )

print("Folder path:", data_folder_path)
```

```python
# Create a txt file
import os

os.makedirs(data_folder_path, exist_ok= True)
```

```python
## Load text file

### TextLoader
a = os.path.join( data_folder_path,"text_files")
print(a)
from langchain_community.document_loaders import TextLoader

loader = TextLoader(os.path.join( data_folder_path,"text_files","machine_learning.txt"), encoding="utf-8")
document = loader.load()

print(document)
```

```python
## Directory loder

from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    os.path.join(data_folder_path,"text_files"),
    glob = "**/*.txt", ## Match patter in file folder
    loader_cls= TextLoader, # load class to be used
    loader_kwargs={'encoding': 'utf-8'},
    show_progress=False
)

all_docs = dir_loader.load()
# all_docs
```

```python
## Pdf Loader
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
from langchain_community.document_loaders import DirectoryLoader

pdf_loader = DirectoryLoader(
    os.path.join(data_folder_path,"pdf_files"),
    glob = "**/*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress= False
)

pdf_docs = pdf_loader.load()
# pdf_docs
```
